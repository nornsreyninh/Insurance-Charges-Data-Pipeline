```python
import pandas as pd
import numpy as np
df= pd.read_csv(r"C:\Users\User\Pictures\insurance.csv")
```

```python
print(df.shape)
print(df.dtypes)
print(df.isna().sum())
print(df.duplicated().sum())
```

```python
df.columns=[c.strip().lower() for c in df.columns]
```

```python
df["sex"]=df["sex"].str.strip().str.lower()
df["smoker"] = df["smoker"].str.strip().str.lower()
df["region"] = df["region"].str.strip().str.lower()
```

```python
assert set(df["sex"]) <= {"male", "female"}
assert set(df["smoker"]) <= {"yes", "no"}
assert set(df["region"]) <= {"southwest", "southeast", "northwest", "northeast"}
```

```python
# Numeric sanity checks
df = df[(df["age"] >= 18) & (df["age"] <= 100)]
df = df[(df["bmi"] > 10) & (df["bmi"] < 60)]
df = df[df["children"] >= 0]
df = df[df["charges"] > 0]
```

```python
before = len(df)
df = df.drop_duplicates()
print(f"Removed {before - len(df)} duplicate rows")
```

```python
def bmi_cat(b):
    if b < 18.5: return "Underweight"
    elif b < 25: return "Normal"
    elif b < 30: return "Overweight"
    else: return "Obese"
df["bmi_category"] = df["bmi"].apply(bmi_cat)
print( df["bmi_category"].value_counts())
```

```python
df["age_group"] = pd.cut(
    df["age"], bins=[17, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56+"]
)
print(df["age_group"].value_counts())
# Flags
```

```python
# Log target (charges are right-skewed)
df["log_charges"] = np.log1p(df["charges"])
```

```python
df.to_csv("insurance_clean.csv", index=False)
print("Clean shape:", df.shape)
```
